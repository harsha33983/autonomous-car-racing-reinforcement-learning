### 3. Reward Distribution Analysis
Visualizing distribution of episodic returns across 100 independent test seeds.

```python
dist_df = pd.read_csv('../results/evaluation/combined_eval_episodes.csv')
plt.figure(figsize=(10, 5))
for algo in ['Random', 'DQN', 'PPO']:
    subset = dist_df[dist_df['algorithm'] == algo]['reward']
    plt.hist(subset, bins=20, alpha=0.5, label=algo)
plt.axvline(200, color='red', linestyle='--', label='Solved (+200)')
plt.title('Reward Distribution Comparison across 100 Episodes')
plt.xlabel('Episodic Return')
plt.ylabel('Count')
plt.legend()
plt.show()
```
